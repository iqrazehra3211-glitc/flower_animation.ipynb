```python
import turtle
import random
import math
import time

# --- 1. Setup ---
screen = turtle.Screen()
screen.setup(900, 700)
screen.bgcolor("#0a0e14")
screen.title("Glow Garden with Crescent Moon")
screen.tracer(0)

t = turtle.Turtle()
t.hideturtle()

# --- 2. Data Structures ---
sparkles = []
for _ in range(60):
    sparkles.append({
        "x": random.randint(-450, 450),
        "y": random.randint(-350, 350),
        "dx": random.uniform(-0.5, 0.5),
        "dy": random.uniform(-0.5, 0.5),
        "size": random.randint(1, 3),
        "phase": random.uniform(0, 10)
    })

flowers = []
for _ in range(12):
    flowers.append({
        "x": random.randint(-380, 380),
        "y": random.randint(-220, 30),
        "size": random.randint(25, 40),
        "offset": random.uniform(0, 5)
    })

# --- 3. Drawing Functions ---

def draw_crescent_moon(x, y, radius):
    """Draws a sharp crescent moon by overlapping two circles"""
    # 1. Main Moon Body
    t.penup()
    t.goto(x, y - radius)
    t.color("#fdfde0") 
    t.begin_fill()
    t.circle(radius)
    t.end_fill()

    # 2. The 'Cutter' (Match background color to create crescent)
    t.penup()
    t.goto(x - (radius * 0.5), y - (radius * 0.8)) 
    t.color("#0a0e14") 
    t.begin_fill()
    t.circle(radius)
    t.end_fill()

    # 3. Static stars around the moon
    t.color("#fdfde0")
    for _ in range(3):
        t.penup()
        t.goto(x + random.randint(-60, 20), y + random.randint(-60, 60))
        t.dot(random.randint(1, 2))

def draw_leaf(x, y, angle, size, sway):
    t.penup()
    t.goto(x, y)
    t.setheading(angle + sway)
    t.color("#0e2f1d")
    t.begin_fill()
    t.circle(size, 90)
    t.left(90)
    t.circle(size, 90)
    t.end_fill()

def draw_flower(x, y, scale, sway):
    # Stem
    t.penup()
    t.goto(x, -350)
    t.pendown()
    t.pensize(2)
    t.color("#1a3d16")
    t.goto(x + sway * 2, y)
    
    # Petals
    t.setheading(90 + sway)
    t.color("#a0ffe3")
    for _ in range(6):
        t.begin_fill()
        for _ in range(2):
            t.circle(scale, 70)
            t.left(110)
        t.end_fill()
        t.left(60)
    
    # Shine Center
    t.penup()
    t.goto(t.xcor(), t.ycor())
    t.color("#ffffff")
    t.dot(scale/2.5)

# --- 4. Animation Loop ---
frame = 0
running = True

def safe_exit():
    global running
    running = False

screen.getcanvas().winfo_toplevel().protocol("WM_DELETE_WINDOW", safe_exit)

while running:
    try:
        t.clear()
        
        # 1. Background Elements (Moon stays still in the corner)
        draw_crescent_moon(350, 250, 45)
        
        # 2. Physics & Wind
        wind = math.sin(frame * 0.05) * 4
        
        # 3. Grass/Leaves
        for i in range(-500, 600, 100):
            draw_leaf(i, -360, 45, 110, wind)
            draw_leaf(i + 30, -360, 65, 85, wind * 1.2)

        # 4. Moving Flowers
        for f in flowers:
            f_pulse = f["size"] + math.sin(frame * 0.08 + f["offset"]) * 3
            f_sway = wind + math.sin(frame * 0.12 + f["offset"]) * 2
            draw_flower(f["x"], f["y"], f_pulse, f_sway)

        # 5. Fireflies
        for s in sparkles:
            s["x"] += s["dx"]
            s["y"] += s["dy"]
            blink = s["size"] + math.sin(frame * 0.2 + s["phase"]) * 1.2
            
            # Wrap around screen
            if abs(s["x"]) > 450: s["x"] *= -0.9
            if abs(s["y"]) > 350: s["y"] *= -0.9
            
            t.penup()
            t.goto(s["x"], s["y"])
            t.dot(max(1, blink), "#fdfd96")

        screen.update()
        frame += 1
        time.sleep(0.01)
        
    except:
        break

print("Garden Finished.")
```
